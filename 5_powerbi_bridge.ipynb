```python
# CELL: Import Required Libraries
#-----------------------------------------------------------

from dotenv import load_dotenv
import os
```

```python
# CELL: Load Environment Variables
#-----------------------------------------------------------

# 1. Load environment variables from file
load_dotenv()

# 2. Assign storage acount name and key vars from env
storage_account_name = os.getenv("STORAGE_ACCOUNT_NAME")
storage_account_key = os.getenv("STORAGE_ACCOUNT_KEY")

# 3. Verify env read successfully
print(f"storage_account_name = {storage_account_name}")
```

```python
# CELL: Configure spark settings
#-----------------------------------------------------------

# 1. Connect to storage account
spark.conf.set(
    f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net",
    storage_account_key,
)

print("Configuration loaded & Auto-create enabled")
```

```python
# CELL: Assign constants
#-----------------------------------------------------------

# 1. Assign  Gold
GOLD = "gold"

# 2. Assign medallion path
GOLD_PATH = f"abfss://{GOLD}@{storage_account_name}.dfs.core.windows.net/"

# 3. Assign datasets
GOLD_DS_SALES_OVERVIEW = "sales_overview"
GOLD_DS_AREA_STATS = "area_stats"
GOLD_DS_SUPPLY_STATS = "supply_stats"
GOLD_DSS = [
    GOLD_DS_SALES_OVERVIEW, 
    GOLD_DS_AREA_STATS, 
    GOLD_DS_SUPPLY_STATS
]
GOLD_TBL_ALIASES = {ds_name: f"{GOLD}_{ds_name}" for ds_name in GOLD_DSS}
```

```python
for gold_ds_name, gold_tbl_alias in GOLD_TBL_ALIASES.items():
    print(f"Saving dataset {gold_ds_name} as table {gold_tbl_alias}")

    gold_ds_path = f"{GOLD_PATH}{gold_ds_name}"
    spark.read.format("delta").load(gold_ds_path).write.mode("overwrite").saveAsTable(gold_tbl_alias)

print("Power BI Bridge: All 3 Tables Registered Successfully")
```
